```python
import time

def download():
    time.sleep(4)
    return "file downlaod"

def install():
    time.sleep(4)
    return "file installed"

print(download())
print(install())
```

```python
import asyncio

async def download():
    await asyncio.sleep(4)
    return "file downlaod"

async def install():
    await asyncio.sleep(2)
    return "file installed"

async def main():
    await asyncio.gather(download(), install())

asyncio.run(main())
```
